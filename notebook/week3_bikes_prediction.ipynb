```python
def united_bike_rentals(
    direct_pickup_bike_rentals: pd.DataFrame, 
    registered_bike_rentals: pd.DataFrame
) -> pd.DataFrame:
    """Merges direct and registered bike rentals into a single master tracking dataset."""
    
    merged_df = pd.merge(
        direct_pickup_bike_rentals,
        registered_bike_rentals,
        on=['datetime', 'is_weekend'],
        how='outer'
    )
    
    # Clean and sum
    merged_df['direct_count'] = merged_df['direct_count'].fillna(0)
    merged_df['registered_count'] = merged_df['registered_count'].fillna(0)
    merged_df['total_rentals'] = merged_df['direct_count'] + merged_df['registered_count']
    
    # Sort chronologically to make downstream EDA and modeling easier
    merged_df = merged_df.sort_values('datetime').reset_index(drop=True)
    
    return merged_df
```

```python
def united_bikes_holidays(
    united_bike_rentals: pd.DataFrame, 
    hollidays: pd.DataFrame
) -> pd.DataFrame:
    """Merges direct and registered bike rentals into a single master tracking dataset."""
    
    merged_df = pd.merge(
        united_bike_rentals,
        hollidays,
        on=['datetime', 'is_weekend'],
        how='outer'
    )
    
    # Clean and sum
    merged_df['direct_count'] = merged_df['direct_count'].fillna(0)
    merged_df['registered_count'] = merged_df['registered_count'].fillna(0)
    merged_df['total_rentals'] = merged_df['direct_count'] + merged_df['registered_count']
    
    # Sort chronologically to make downstream EDA and modeling easier
    merged_df = merged_df.sort_values('datetime').reset_index(drop=True)
    
    return merged_df
```

```python
def merge_bikes_with_holidays(bike_df, holiday_df):
    # 1. Create matching date-only keys for both dataframes
    bike_df['date_key'] = bike_df['datetime'].dt.date
    holiday_df['date_key'] = holiday_df['datetime'].dt.date
    print(bike_df.head())
    print("================")
    print(holiday_df.head())
    print("================")
    holiday_subset = holiday_df[['date_key', 'is_holiday']]
    # 3. Perform a left join using our normalized date key
    merged_df = pd.merge(bike_df, holiday_subset, on='date_key', how='left')
    print(merged_df.head())
    print("================")
    
    # 4. Fill the unfulfilled NaN spaces with False
    merged_df['is_holiday'] = merged_df['is_holiday'].fillna(False)
    print(merged_df.head())
    print("================")
    
    # 5. Drop the temporary key so it doesn't clutter our final output
    merged_df = merged_df.drop(columns=['date_key'])
    print(merged_df.head())
    print("================")
    return merged_df
```

```python
def verify_holiday_broadcasting(merged_df, target_date_string):
    # Convert string to a date object to match against the datetime column
    target_date = pd.to_datetime(target_date_string).date()
    
    # Filter for rows matching only that specific calendar day
    day_rows = merged_df[merged_df['datetime'].dt.date == target_date]
    
    # Select only the relevant columns to display
    verification_subset = day_rows[['datetime', 'is_holiday']]
    
    return verification_subset
```

```python
def dfs_united(bikes_hollidays, weather):
    # 1. Create matching date-only keys for both dataframes
    """Merges direct and registered bike rentals into a single master tracking dataset."""
    
    bikes_hollidays = bikes_hollidays.sort_values('datetime')
    weather = weather.sort_values('datetime')
    # how='left': This ensures every single hour from bike_df is preserved. If weather_df lacks a matching datetime,
    merged_df = pd.merge(bikes_hollidays, weather, on='datetime', how='left')
    
    return merged_df.reset_index(drop=True)
```

```python

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

def run_advanced_baseline(df: pd.DataFrame):
    # 1. Create a copy so we don't accidentally mutate the original data in memory
    model_df = df.copy()
    
    # 2. Convert the boolean switches to 1s and 0s directly in the dataframe
    boolean_features = ['is_weekend', 'is_holiday'] # (Ensure this exactly matches your column name)
    model_df[boolean_features] = model_df[boolean_features].astype(int)
    
    # 3. Dismantle datetime into pure numbers before the model sees it
    model_df['hour'] = model_df['datetime'].dt.hour
    model_df['month'] = model_df['datetime'].dt.month
    
    # 4. Define the feature matrix (X) by dropping targets AND the raw datetime
    columns_to_drop = ['datetime', 'direct_count', 'registered_count', 'total_rentals']
    X = model_df.drop(columns=columns_to_drop)
    
    # 5. Extract the target vector (y)
    y = model_df['total_rentals']
    
    # 6. Split the data chronologically
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
    
    # 7. Train the mathematical model
    baseline_model = LinearRegression()
    baseline_model.fit(X_train, y_train)
    
    # 8. Evaluate
    predictions = baseline_model.predict(X_test)
    rmse = root_mean_squared_error(y_test, predictions)
    
    print(f"Advanced Baseline RMSE: {rmse:.2f}")
    return baseline_model
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score

def run_full_linear_model(df: pd.DataFrame):
    # 1. Create a copy
    model_df = df.copy()
    
    # 2. Convert boolean switches
    boolean_features = ['is_weekend', 'is_holiday'] 
    model_df[boolean_features] = model_df[boolean_features].astype(int)
    
    # 3. Dismantle datetime
    model_df['hour'] = model_df['datetime'].dt.hour
    model_df['month'] = model_df['datetime'].dt.month
    
    # 4. Define continuous features that NEED scaling
    # (Adjust these names if your weather columns are named differently)
    continuous_features = ['hour', 'month', 'temperature_c', 'humidity', 'windspeed_kmh', 'conditions', 'perceived_temperature_c']
    
    # Apply the scaler ONLY to the continuous columns
    scaler = StandardScaler()
    model_df[continuous_features] = scaler.fit_transform(model_df[continuous_features])
    
    # 5. Define X and y
    columns_to_drop = ['datetime', 'direct_count', 'registered_count', 'total_rentals']
    X = model_df.drop(columns=columns_to_drop)
    y = model_df['total_rentals']
    
    # 6. Split chronologically
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False, )
    
    # 7. Train the mathematical model
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    # 8. Evaluate
    predictions = model.predict(X_test)
    rmse = root_mean_squared_error(y_test, predictions)
    mae = mean_absolute_error(y_test, predictions)
    r2 = r2_score(y_test, predictions)
    
    print("=== Full Linear Model Evaluation ===")
    print(f"RMSE: {rmse:.2f} bikes")
    print(f"MAE:  {mae:.2f} bikes")
    print(f"R²:   {r2:.2f}")
    
    return model, scaler
```

```python
from sklearn.inspection import permutation_importance

def analyze_feature_importance(model, X_test: pd.DataFrame, y_test: pd.Series):
    print("Scrambling columns to calculate exact feature reliance...")
    
    # 1. Run the permutation test
    result = permutation_importance(
        model, 
        X_test, 
        y_test, 
        scoring='neg_root_mean_squared_error', 
        n_repeats=10, 
        random_state=42
    )
    
    # 2. Package the math into a readable table
    importance_df = pd.DataFrame({
        'Feature': X_test.columns,
        'Importance_Score': result.importances_mean
    })
    
    # 3. Sort from most critical to least critical
    importance_df = importance_df.sort_values(by='Importance_Score', ascending=False)
    importance_df = importance_df.reset_index(drop=True)
    
    print("\n=== Feature Importance Ranking ===")
    print(importance_df)
    
    return importance_df
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score

def run_advanced_tree_model(df: pd.DataFrame):
    # 1. Create a copy
    model_df = df.copy()
    
    # 2. Convert boolean switches to integers
    boolean_features = ['is_weekend', 'is_holiday'] 
    model_df[boolean_features] = model_df[boolean_features].astype(int)
    
    # 3. Dismantle datetime
    model_df['hour'] = model_df['datetime'].dt.hour
    model_df['month'] = model_df['datetime'].dt.month
    
    # Notice what is MISSING here: No StandardScaler! 
    # Tree models do not care if humidity is 80 and is_weekend is 1.
    
    # 4. Define X and y (Keeping ALL weather columns, even the ones with NaNs)
    columns_to_drop = ['datetime', 'direct_count', 'registered_count', 'total_rentals']
    X = model_df.drop(columns=columns_to_drop)
    y = model_df['total_rentals']
    
    # 5. Split chronologically
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
    
    # 6. Train the Gradient Boosting Model
    # max_iter=100 means it will build 100 sequential decision trees
    tree_model = HistGradientBoostingRegressor(max_iter=200)
    tree_model.fit(X_train, y_train)
    
    # 7. Evaluate
    predictions = tree_model.predict(X_test)
    rmse = root_mean_squared_error(y_test, predictions)
    mae = mean_absolute_error(y_test, predictions)
    r2 = r2_score(y_test, predictions)
    
    analyze_feature_importance(tree_model, X_test, y_test)
    
    print("=== Advanced Tree Model Evaluation ===")
    print(f"RMSE: {rmse:.2f} bikes")
    print(f"MAE:  {mae:.2f} bikes")
    print(f"R²:   {r2:.2f}")
    
    return tree_model
```

```python
def engineer_time_features(df: pd.DataFrame):
    # 1. Create a working copy and ensure it is perfectly sorted by time
    # If the rows are out of order, the shift math will be entirely wrong.
    df_engineered = df.copy()
    df_engineered = df_engineered.sort_values('datetime').reset_index(drop=True)
    
    # 2. Lag Feature: Shift the answers down by exactly 24 rows (hours)
    df_engineered['rentals_24h_ago'] = df_engineered['total_rentals'].shift(24)
    
    # 3. Rolling Feature: Calculate the mean of the 3 strictly prior hours
    # .shift(1) ensures we don't accidentally include the CURRENT hour's rentals (which would be cheating)
    df_engineered['rolling_avg_3h'] = df_engineered['total_rentals'].shift(1).rolling(window=3).mean()
    
    # 4. Clean up the unavoidable NaNs
    # Shifting data down creates 24 blank rows at the very top of your dataset.
    df_engineered = df_engineered.dropna().reset_index(drop=True)
    
    return df_engineered
```
